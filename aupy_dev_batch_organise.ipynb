# Setting Paths and Arranging the Data

Data from the ROCC comes in various directory structures, and with the filenames relabelled.

We need to reorganise and relabel this data, which can all be done via accessing the xml files attached with each image.

In steps:

1. Access directory for all data from the given sol.
2. Find all image data (as png's) within the sol directory.
3. Find the paired xml files for each image.
4. Extract filenames and other organisational metadata from the images and xml files.
5. Put the data into a new processing directory for that sol.

It seems that only WAC's ad some HRC have the metadata we are looking for, so we're going to only get images with HRC or WAC in the name (i.e. ignoring Nav and Loc Cams).

```python
from pathlib import Path
```

Choose a Sol to process.

Let's develop on Sol39, because we have a range of cameras and scenes on this sol.

```python
sol_dir = Path('/Users/sci/Documents/Documents - Roger’s MacBook Air/projects/esaskp_2025-2027/field_trials/SKP_FT1/data/downlink/sol39')
```

I'm keeping downloaded data in a 'data/downlink' directory.

I then am putting processing steps in a 'data/processing' directory.

We need to define the path to that processing directory.

```python
pr_dir = sol_dir.parent.parent / 'processing'
```

These lines, calling aupy, does the steps listed above, and the 'organise' function executes the file copying and renaming into the new directory tree in the processing folder.

The processed products are put under 'data/processing/solXX/aupy/' to distinguish this set of products from others.

```python
sol_parser = aupy.ROCCDownlinkDirTool(sol_dir, pr_dir)
new_png_xml_pairs, out_sol_dir = sol_parser.organise()
```

We can see which PanCam images have been skipped, because they don't have xml files associated with them.

This gives us the new list of renamed pngs and xmls, and the directory for the sol that they are held in.

The idea is that the new sol directory can be passed to other aupy processes for performing colour and multispectral calibration.

```python
out_sol_dir
```

We can also batch process these.

Batch process all downloaded sols:

```python
data_dir = Path('/Users/sci/Documents/Documents - Roger’s MacBook Air/projects/esaskp_2025-2027/field_trials/SKP_FT1/data/downlink')
```

```python
# make a list of all the directories in the data directory
sols = [d.name for d in data_dir.iterdir()]
# drop .DS_Store
sols = [s for s in sols if s != '.DS_Store']
sols.sort()
```

```python
sols
```

```python
for sol in sols:
    sol_dir = data_dir / sol
    sol_parser = aupy.ROCCDownlinkDirTool(sol_dir, pr_dir)
    new_png_xml_pairs, out_sol_dir = sol_parser.organise()
    print(f"Processed sol {sol}, output directory: {out_sol_dir}")
```
